Text Cleaning

```python
df['text']=df['text'].apply(lambda x: x.lower())
```

```python
df.head()
```

removing Punctiation

```python
import string
def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df["text"]=df["text"].apply(remove_punc)
```

Removing numbers

```python
def remove_numbers(txt):
  new=""
  for i in txt:
    if not i.isdigit():
      new=new+i
  return new
  df["text"]=df["text"].apply(removing_numbers)
```

Remove Emojis

```python
def remove_emojis(txt):
  new=""
  for i in txt:
    if i.isascii():
      new=new+i
      return new
      df["text"]=df["text"].apply(remove_emojis)
```

Remove Stopwords

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download("punkt")
nltk.download("stopwords")
# punkt is a pre-trained sentence and word tokeizer. It helps NLTK split text into:

# Sentences
# Words

# It uses language-specific punctuation rules to perform accurate tokenization.
```

```python
stop_words=set(stopwords.words("english"))
stop_words
```

```python
len(stop_words)
```

```python
df.loc[1]["text"]
```

```python
# now remove stop words from text
def remove (txt):
  # words=word_tokenize(txt)
  words=txt.split()
  cleaned=[]
  for i in words:
    if not i in stop_words:
      cleaned.append(i)
  return" ".join(cleaned)
```

```python
df["text"]=df["text"].apply(remove)
```

```python
def preprocess_text(text):
    text = text.lower()
    text = remove_punc(text) # Assuming remove_punc is defined in previous cells
    text = remove(text) # Assuming remove is defined in previous cells
    return text

def predict_sentiment(user_input):
    processed_input = preprocess_text(user_input)
    input_vectorized = tfdf_vectorize.transform([processed_input])
    predicted_emotion_id = lr_model.predict(input_vectorized)[0]

    # Reverse the emotion_numbers dictionary to get the name from the ID
    id_to_emotion = {v: k for k, v in emotion_numbers.items()}
    predicted_emotion_name = id_to_emotion.get(predicted_emotion_id, "Unknown")

    print(f"User Input: {user_input}")
    print(f"Predicted Sentiment: {predicted_emotion_name}")

# Example usage:
user_text = input("Enter your text: ")
predict_sentiment(user_text)
```
